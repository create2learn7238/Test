```python
import pandas as pd
df = pd.read_csv("IMDB-Dataset.csv")
print(df)
```

```python
def Analyze(dataset):
    print("Size",dataset.size)
    print("Columns",dataset.columns)
    print("DataTypes")
    print(dataset.dtypes)
    print("++++++++++++++++++++++++++++")
    print("Missing in review ",dataset["review"].isna().sum())
    print("Missing in sentiment",dataset["sentiment"].isna().sum())
    print("++++++++++++++++++++++++++++")
    df = dataset[(dataset["review"].str.len() >= 1000) | (dataset["review"].str.len() <= 300)]
    print("removed duplication ... ")
Analyze(df)
```

```python
def check(val):
    if val=="positive":
        return 3
    elif val=="neutral":
        return 2
    elif val=="negative":
        return 1
    else :
        return 0
df["rate"]=df["sentiment"].apply(check)
print("done")
```

```python
y=df["rate"]
x=df.drop(columns=["rate"])
from sklearn.model_selection import train_test_split
x_train, x_test,y_train ,y_test=train_test_split(x,y,random_state=42)
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score
model_SVC=SVC(C=1.0)
model_SVC.fit(x_train,y_train)
y_pred=model_SVC.predict(x_test)
```
